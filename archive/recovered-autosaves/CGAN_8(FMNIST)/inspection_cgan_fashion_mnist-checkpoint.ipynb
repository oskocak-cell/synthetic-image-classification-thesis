```python
# Cell 1 — Imports & Paths

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from glob import glob

PROJECT_DIR  = '/path/to/local-user/Desktop/CGAN_8(FMNIST)'
SAVE_DIR     = os.path.join(PROJECT_DIR, 'inspection_fashion_mnist', 'cgan', 'fashion_mnist', 'seed_42')
SYNTH_PATH   = os.path.join(PROJECT_DIR, 'data_fashion', 'synthetic', 'seed_42', 'cgan', 'fashion_mnist', 'fashion_mnist_synthetic.npz')
REAL_PATH    = os.path.join(PROJECT_DIR, 'data', 'processed', 'seed_42', 'fashion_mnist.npz')
MODEL_DIR    = os.path.join(PROJECT_DIR, 'models_fashion', 'cgan', 'fashion_mnist', 'seed_42')
SEARCH_DIR   = os.path.join(MODEL_DIR, 'search')
FINAL_DIR    = os.path.join(MODEL_DIR, 'final')

os.makedirs(SAVE_DIR, exist_ok=True)

CLASSES = 10
SEED    = 42
```

```python
# Cell 2 — Load synthetic data & print summary

CLASS_NAMES = [
    'T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]

data    = np.load(SYNTH_PATH)
X_train = data['X_train']
y_train = data['y_train']
X_val   = data['X_val']
y_val   = data['y_val']

print('── Keys ──────────────────────────────────')
print(list(data.keys()))

print('\n── Shapes ────────────────────────────────')
for k in ['X_train', 'y_train', 'X_val', 'y_val']:
    print(f'  {k}: {data[k].shape}')

print('\n── Dtypes ────────────────────────────────')
for k in ['X_train', 'y_train', 'X_val', 'y_val']:
    print(f'  {k}: {data[k].dtype}')

print('\n── Pixel range (ALL splits) ──────────────')
print(f'  X_train: [{X_train.min()}, {X_train.max()}]')
print(f'  X_val:   [{X_val.min()}, {X_val.max()}]')

print('\n── Class distribution ────────────────────')
for split_name, y in [('y_train', y_train), ('y_val', y_val)]:
    counts = np.bincount(y, minlength=CLASSES)
    print(f'\n  {split_name}  (n={len(y)}):')
    for cls, cnt in enumerate(counts):
        print(f'    {CLASS_NAMES[cls]}: {cnt}')

# Synthetic vs real comparison — train
print('\n── Synthetic vs Real class counts (train) ─')
real             = np.load(REAL_PATH)
y_real_tr        = real['y_train']
y_real_val       = real['y_val']
counts_synth_tr  = np.bincount(y_train,    minlength=CLASSES)
counts_real_tr   = np.bincount(y_real_tr,  minlength=CLASSES)
counts_synth_val = np.bincount(y_val,      minlength=CLASSES)
counts_real_val  = np.bincount(y_real_val, minlength=CLASSES)

print(f"  {'Class':<15} {'Real Train':>12} {'Synth Train':>12} {'Match':>8}")
print(f"  {'-'*52}")
for cls in range(CLASSES):
    match = '\u2713' if counts_synth_tr[cls] == counts_real_tr[cls] else '~'
    print(f"  {CLASS_NAMES[cls]:<15} {counts_real_tr[cls]:>12} {counts_synth_tr[cls]:>12} {match:>8}")

# Synthetic vs real comparison — val
print(f'\n── Synthetic vs Real class counts (val) ───')
print(f"  {'Class':<15} {'Real Val':>12} {'Synth Val':>12} {'Match':>8}")
print(f"  {'-'*52}")
for cls in range(CLASSES):
    match = '\u2713' if counts_synth_val[cls] == counts_real_val[cls] else '~'
    print(f"  {CLASS_NAMES[cls]:<15} {counts_real_val[cls]:>12} {counts_synth_val[cls]:>12} {match:>8}")
```

```python
# Cell 3 — Class distribution bar chart

counts_tr = np.bincount(y_train, minlength=CLASSES)

fig, ax = plt.subplots(figsize=(14, 5))
ax.bar(range(CLASSES), counts_tr, color='steelblue')
ax.set_xlabel('Class')
ax.set_ylabel('Number of samples')
ax.set_title('FashionMNIST cGAN — Synthetic Class Distribution')
ax.set_xticks(range(CLASSES))
ax.set_xticklabels(CLASS_NAMES, rotation=45, ha='right')
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, 'class_distribution_synthetic.png'),
            dpi=150, bbox_inches='tight')
plt.show()
print('Saved: class_distribution_synthetic.png')
```

```python
# Cell 4 — 5 sample images per class from synthetic X_train (fixed spacing)
# FashionMNIST is grayscale (28x28x1): squeeze channel dim, use cmap='gray'

fig, axes = plt.subplots(CLASSES, 5, figsize=(12, CLASSES * 2.2))
for class_idx in range(CLASSES):
    class_mask   = y_train == class_idx
    class_images = X_train[class_mask]
    indices      = np.linspace(0, len(class_images) - 1, 5, dtype=int)
    for col, img_idx in enumerate(indices):
        axes[class_idx, col].imshow(class_images[img_idx].squeeze(), cmap='gray')
        axes[class_idx, col].axis('off')
    axes[class_idx, 0].set_title(CLASS_NAMES[class_idx], fontsize=9,
                                  loc='left', pad=4)

plt.suptitle('Synthetic X_train — 5 Samples per Class (cGAN FashionMNIST)', fontsize=12)
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, 'synthetic_samples_per_class.png'),
            dpi=150, bbox_inches='tight')
plt.show()
print('Saved: synthetic_samples_per_class.png')
```

```python
# Cell 4b — Additional random sample grid for visual verification
# Different seed from SEED=42 so different samples each run.

fig, axes = plt.subplots(CLASSES, 5, figsize=(12, CLASSES * 2.2))
rng = np.random.default_rng()  # no seed — different each run

data      = np.load(SYNTH_PATH)
X_train   = data['X_train']
y_train   = data['y_train']
CLASS_NAMES = [
    'T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]

for class_idx in range(CLASSES):
    class_mask   = y_train == class_idx
    class_images = X_train[class_mask]
    indices      = rng.choice(len(class_images), size=5, replace=False)
    for col, img_idx in enumerate(indices):
        axes[class_idx, col].imshow(class_images[img_idx].squeeze(), cmap='gray')
        axes[class_idx, col].axis('off')
    axes[class_idx, 0].set_title(CLASS_NAMES[class_idx], fontsize=9,
                                  loc='left', pad=4)

plt.suptitle('Synthetic X_train — Random Samples per Class (cGAN FashionMNIST)', fontsize=12)
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, 'synthetic_samples_per_class_random.png'),
            dpi=150, bbox_inches='tight')
plt.show()
print('Saved: synthetic_samples_per_class_random.png')
```

```python
# Cell 5 — Image grids from final training
grid_paths = sorted(glob(os.path.join(FINAL_DIR, 'epoch_*', 'generated_epoch_*.png')))
print(f'Found {len(grid_paths)} image grids from final training:')
for p in grid_paths:
    print(f'  {os.path.basename(os.path.dirname(p))}/{os.path.basename(p)}')

if grid_paths:
    n_grids = len(grid_paths)
    n_cols  = 3
    n_rows  = int(np.ceil(n_grids / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 5, n_rows * 6))
    axes = np.array(axes).flatten()

    for i, (ax, path) in enumerate(zip(axes, grid_paths)):
        img = mpimg.imread(path)
        ax.imshow(img)
        ax.axis('off')
        ax.set_title(os.path.basename(os.path.dirname(path)), fontsize=10)

    for ax in axes[n_grids:]:
        ax.axis('off')

    plt.suptitle('Generated Image Grids — cGAN FashionMNIST Final Training Progress', fontsize=13)
    plt.tight_layout()
    plt.savefig(os.path.join(SAVE_DIR, 'final_training_image_grids.png'),
                dpi=150, bbox_inches='tight')
    plt.show()
    print('Saved: final_training_image_grids.png')
else:
    print('No image grids found — check FINAL_DIR path.')
```

```python
# Cell 7 — FID vs Epoch: All search runs

fig, ax = plt.subplots(figsize=(12, 6))
for _, row in search_df.iterrows():
    fid_path = os.path.join(SEARCH_DIR, row['run_name'], 'fid_scores.csv')
    if not os.path.exists(fid_path):
        continue
    fid_df = pd.read_csv(fid_path)
    ax.plot(fid_df['epoch'], fid_df['fid'], marker='o', markersize=4,
            label=row['run_name'])

ax.set_xlabel('Epoch')
ax.set_ylabel('FID Score')
ax.set_title('FID vs Epoch — All Search Runs (cGAN FashionMNIST)')
ax.legend(fontsize=8, loc='upper right')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, 'fashion_mnist_search_fid.png'),
            dpi=150, bbox_inches='tight')
plt.show()
print('Saved: fashion_mnist_search_fid.png')
```

```python
# Cell L1 — L1 Nearest Neighbor Distance (synthetic train vs real train)
# For each class: sample N_SAMPLES synthetic images, find closest real
# training image by L1 distance within the same class.
# Demonstrates the model is generalizing, not memorizing.
# Reference: Foster, D. (2023). Generative Deep Learning (2nd ed.). O'Reilly.
# FashionMNIST is grayscale: squeeze channel dim, use cmap='gray' for display.
SEED      = 42
N_SAMPLES = 5

def l1_distance(img1, img2):
    return np.mean(np.abs(img1.astype('float32') - img2.astype('float32')))

real_data = np.load(REAL_PATH)
X_real_tr = real_data['X_train']
y_real_tr = real_data['y_train']
l1_results = []

rng = np.random.default_rng(SEED)

for class_idx, class_name in enumerate(CLASS_NAMES):
    synth_mask     = y_train == class_idx
    synth_images   = X_train[synth_mask]
    sample_indices = rng.choice(len(synth_images), size=N_SAMPLES, replace=False)
    sampled_synth  = synth_images[sample_indices]

    real_mask   = y_real_tr == class_idx
    real_images = X_real_tr[real_mask]

    class_l1_distances = []
    nearest_reals      = []

    for synth_img in sampled_synth:
        distances = [l1_distance(synth_img, real_img) for real_img in real_images]
        min_idx   = np.argmin(distances)
        min_dist  = distances[min_idx]
        nearest_reals.append((real_images[min_idx], min_dist))
        class_l1_distances.append(min_dist)

    # One figure per class — 2 rows (synthetic / real), N_SAMPLES columns
    fig, axes = plt.subplots(2, N_SAMPLES, figsize=(N_SAMPLES * 3, 7))
    fig.suptitle(f'Class: {class_name}', fontsize=16, fontweight='bold', y=1.01)

    for col in range(N_SAMPLES):
        axes[0, col].imshow(sampled_synth[col].squeeze(), cmap='gray')
        axes[0, col].axis('off')
        if col == 0:
            axes[0, col].set_ylabel('Synthetic', fontsize=13, fontweight='bold', labelpad=10)

    for col, (nearest, dist) in enumerate(nearest_reals):
        axes[1, col].imshow(nearest.squeeze(), cmap='gray')
        axes[1, col].axis('off')
        axes[1, col].set_title(f'L1 = {dist:.3f}', fontsize=11, pad=6)
        if col == 0:
            axes[1, col].set_ylabel('Nearest Real', fontsize=13, fontweight='bold', labelpad=10)

    # Re-enable y-axis so ylabel is visible
    for row in range(2):
        axes[row, 0].axis('on')
        axes[row, 0].tick_params(left=False, bottom=False,
                                  labelleft=False, labelbottom=False)
        for spine in axes[row, 0].spines.values():
            spine.set_visible(False)

    plt.tight_layout()
    fname = f"l1_{class_name.lower().replace(' ', '_').replace('/', '-')}.png"
    plt.savefig(os.path.join(SAVE_DIR, fname), dpi=150, bbox_inches='tight')
    plt.show()
    print(f'Saved: {fname}')

    mean_class_l1 = np.mean(class_l1_distances)
    n_synth       = int(np.sum(synth_mask))
    l1_results.append({
        'class':            class_name,
        'n_samples':        n_synth,
        'mean_l1_distance': round(mean_class_l1, 4)
    })
    print(f'  {class_name:<15} mean L1: {mean_class_l1:.4f}  (n={n_synth})')

# Weighted overall mean
weighted_sum     = sum(r['mean_l1_distance'] * r['n_samples'] for r in l1_results)
total_samples    = sum(r['n_samples'] for r in l1_results)
weighted_overall = weighted_sum / total_samples
l1_results.append({
    'class':            'Overall (weighted)',
    'n_samples':        total_samples,
    'mean_l1_distance': round(weighted_overall, 4)
})
print(f'\nOverall weighted mean L1: {weighted_overall:.4f}')

l1_df = pd.DataFrame(l1_results)
l1_df.to_csv(os.path.join(SAVE_DIR, 'l1_distances_fashion_mnist.csv'), index=False)
print('Saved: l1_distances_fashion_mnist.csv')
```
