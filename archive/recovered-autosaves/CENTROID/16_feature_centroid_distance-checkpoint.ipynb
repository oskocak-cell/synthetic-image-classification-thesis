```python
# ── Cell 1: Imports & Parameters ─────────────────────────────────────────────
# Notebook 16: Class-wise Feature Centroid Distance
# Evaluates synthetic data quality by comparing per-class MobileNetV2 feature
# centroids between real (train) and synthetic (train) splits.
# Metric: Mean Centroid Distance (MCD) = mean L2(real_centroid_c, synth_centroid_c)
# Uses already-extracted features from notebooks 13, 14, 15.

import os
import numpy as np

SEED = 42
np.random.seed(SEED)

PROJECT_DIR   = '/content/drive/MyDrive/thesis_project'
FEAT_DIR      = os.path.join(PROJECT_DIR, 'data', 'extracted_features', 'seed_42')
REAL_DATA_DIR = os.path.join(PROJECT_DIR, 'data', 'processed', 'seed_42')
SYNTH_DIR     = os.path.join(PROJECT_DIR, 'data', 'synthetic', 'seed_42')

DATASETS = {
    'fashion_mnist': {'classes': 10},
    'path_mnist':    {'classes': 9},
    'eurosat':       {'classes': 10},
}

MODELS = ['ddim', 'cgan']
```

```python
# ── Cell 2: Helper Functions ──────────────────────────────────────────────────

def load_real_train(dataset_name):
    """Load real train features and labels."""
    feat_path  = os.path.join(FEAT_DIR, 'real', f'{dataset_name}_x_train_features.npy')
    label_path = os.path.join(REAL_DATA_DIR, f'{dataset_name}.npz')
    F = np.load(feat_path)
    y = np.load(label_path)['y_train'].squeeze()
    assert len(F) == len(y), f'Feature/label mismatch for real {dataset_name}: {len(F)} vs {len(y)}'
    return F, y


def load_synth_train(dataset_name, model_name):
    """Load synthetic train features and labels."""
    feat_path  = os.path.join(FEAT_DIR, model_name, f'{dataset_name}_x_train_features.npy')
    label_path = os.path.join(SYNTH_DIR, model_name, dataset_name, f'{dataset_name}_synthetic.npz')
    F = np.load(feat_path)
    y = np.load(label_path)['y_train'].squeeze()
    assert len(F) == len(y), f'Feature/label mismatch for {model_name} {dataset_name}: {len(F)} vs {len(y)}'
    return F, y


def class_centroid_distances(F_real, y_real, F_synth, y_synth, n_classes):
    """
    For each class c, compute L2 distance between real and synthetic centroids.
    Returns array of shape (n_classes,); NaN if a class is missing in either split.
    """
    distances = np.full(n_classes, np.nan)
    for c in range(n_classes):
        real_mask  = (y_real  == c)
        synth_mask = (y_synth == c)
        if real_mask.sum() == 0 or synth_mask.sum() == 0:
            print(f'  Warning: class {c} missing in real or synthetic split — skipping')
            continue
        centroid_real  = F_real[real_mask].mean(axis=0)
        centroid_synth = F_synth[synth_mask].mean(axis=0)
        distances[c]   = np.linalg.norm(centroid_real - centroid_synth)
    return distances
```

```python
# ── Cell 3: Compute Centroid Distances ────────────────────────────────────────

results = {}   # results[model][dataset] = distances array (n_classes,)

for model in MODELS:
    results[model] = {}
    print(f'\n{"="*60}')
    print(f'Model: {model.upper()}')
    print(f'{"="*60}')

    for name, cfg in DATASETS.items():
        n_classes = cfg['classes']
        print(f'\n  Dataset: {name}  (n_classes={n_classes})')

        F_real,  y_real  = load_real_train(name)
        F_synth, y_synth = load_synth_train(name, model)

        dists = class_centroid_distances(F_real, y_real, F_synth, y_synth, n_classes)
        results[model][name] = dists

        for c, d in enumerate(dists):
            tag = f'{d:.4f}' if not np.isnan(d) else 'N/A'
            print(f'    Class {c:2d}: {tag}')
        print(f'  → Mean Centroid Distance (MCD): {np.nanmean(dists):.4f}')
```

```python
# ── Cell 4: Summary Table ─────────────────────────────────────────────────────

print(f'\n{"Dataset":<20} {"DDIM MCD":>12} {"cGAN MCD":>12}')
print('-' * 46)
for name in DATASETS:
    ddim_mcd = np.nanmean(results['ddim'][name])
    cgan_mcd = np.nanmean(results['cgan'][name])
    print(f'{name:<20} {ddim_mcd:>12.4f} {cgan_mcd:>12.4f}')
```

```python
# ── Cell 5: Save Results ──────────────────────────────────────────────────────

OUT_DIR = os.path.join(PROJECT_DIR, 'results', 'feature_centroid_distances')
os.makedirs(OUT_DIR, exist_ok=True)

for model in MODELS:
    for name in DATASETS:
        out_path = os.path.join(OUT_DIR, f'{name}_{model}_centroid_distances.npy')
        np.save(out_path, results[model][name])
        print(f'Saved: {out_path}')

print('\nDone.')

import pandas as pd

rows = []
for model in MODELS:
    for name in DATASETS:
        dists = results[model][name]
        row = {'model': model, 'dataset': name}
        for c, d in enumerate(dists):
            row[f'class_{c}'] = round(float(d), 4) if not np.isnan(d) else None
        row['mcd'] = round(float(np.nanmean(dists)), 4)
        rows.append(row)

df = pd.DataFrame(rows)
csv_path = os.path.join(OUT_DIR, 'centroid_distances_summary.csv')
df.to_csv(csv_path, index=False)
print(f'Saved: {csv_path}')
```
