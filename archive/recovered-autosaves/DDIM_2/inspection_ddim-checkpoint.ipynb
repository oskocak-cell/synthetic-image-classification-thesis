```python
# Cell 1 — Imports & Paths (updated)

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from glob import glob

PROJECT_DIR  = '/path/to/local-user/Desktop/DDIM_2'
SAVE_DIR     = os.path.join(PROJECT_DIR, 'inspection', 'ddim', 'eurosat', 'seed_42')
SYNTH_PATH   = os.path.join(PROJECT_DIR, 'data', 'synthetic', 'seed_42', 'ddim', 'eurosat', 'eurosat_synthetic.npz')
REAL_PATH    = os.path.join(PROJECT_DIR, 'data', 'processed', 'seed_42', 'eurosat.npz')
MODEL_DIR    = os.path.join(PROJECT_DIR, 'models', 'ddim', 'eurosat', 'seed_42')
SEARCH_DIR   = os.path.join(MODEL_DIR, 'search')
FINAL_DIR    = os.path.join(MODEL_DIR, 'final')

os.makedirs(SAVE_DIR, exist_ok=True)

CLASSES= 10
SEED = 42
```

```python
# Cell 2 — Load synthetic data & print summary (Item 1)

CLASS_NAMES = [
    'AnnualCrop', 'Forest', 'HerbaceousVegetation', 'Highway', 'Industrial',
    'Pasture', 'PermanentCrop', 'Residential', 'River', 'SeaLake'
]

data    = np.load(SYNTH_PATH)
X_train = data['X_train']
y_train = data['y_train']
X_val   = data['X_val']
y_val   = data['y_val']

print("── Keys ──────────────────────────────────")
print(list(data.keys()))

print("\n── Shapes ────────────────────────────────")
for k in ['X_train', 'y_train', 'X_val', 'y_val']:
    print(f"  {k}: {data[k].shape}")

print("\n── Dtypes ────────────────────────────────")
for k in ['X_train', 'y_train', 'X_val', 'y_val']:
    print(f"  {k}: {data[k].dtype}")

print("\n── Pixel range (ALL splits) ──────────────")
print(f"  X_train: [{X_train.min()}, {X_train.max()}]")
print(f"  X_val:   [{X_val.min()}, {X_val.max()}]")

print("\n── Class distribution ────────────────────")
for split_name, y in [('y_train', y_train), ('y_val', y_val)]:
    counts = np.bincount(y, minlength=CLASSES)
    print(f"\n  {split_name}  (n={len(y)}):")
    for cls, cnt in enumerate(counts):
        print(f"    {CLASS_NAMES[cls]}: {cnt}")

# Synthetic vs real comparison — train
print("\n── Synthetic vs Real class counts (train) ─")
real            = np.load(REAL_PATH)
y_real_tr       = real['y_train']
y_real_val      = real['y_val']
counts_synth_tr = np.bincount(y_train,    minlength=CLASSES)
counts_real_tr  = np.bincount(y_real_tr,  minlength=CLASSES)
counts_synth_val= np.bincount(y_val,      minlength=CLASSES)
counts_real_val = np.bincount(y_real_val, minlength=CLASSES)

print(f"  {'Class':<25} {'Real Train':>12} {'Synth Train':>12} {'Match':>8}")
print(f"  {'-'*60}")
for cls in range(CLASSES):
    match = '✓' if counts_synth_tr[cls] == counts_real_tr[cls] else '~'
    print(f"  {CLASS_NAMES[cls]:<25} {counts_real_tr[cls]:>12} {counts_synth_tr[cls]:>12} {match:>8}")

# Synthetic vs real comparison — val
print(f"\n── Synthetic vs Real class counts (val) ───")
print(f"  {'Class':<25} {'Real Val':>12} {'Synth Val':>12} {'Match':>8}")
print(f"  {'-'*60}")
for cls in range(CLASSES):
    match = '✓' if counts_synth_val[cls] == counts_real_val[cls] else '~'
    print(f"  {CLASS_NAMES[cls]:<25} {counts_real_val[cls]:>12} {counts_synth_val[cls]:>12} {match:>8}")
```

```python
# Cell 3 — Class distribution bar chart (Item 2)

counts_tr = np.bincount(y_train, minlength=CLASSES)

fig, ax = plt.subplots(figsize=(14, 5))
ax.bar(range(CLASSES), counts_tr, color='steelblue')
ax.set_xlabel('Class')
ax.set_ylabel('Number of samples')
ax.set_title('eurosat synthetic class distribution')
ax.set_xticks(range(CLASSES))
ax.set_xticklabels(CLASS_NAMES, rotation=45, ha='right')
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, 'class_distribution_synthetic.png'),
            dpi=150, bbox_inches='tight')
plt.show()
print("Saved: class_distribution_synthetic.png")
```

```python
# Cell 4 — 5 sample images per class from synthetic X_train (Item 3)

fig, axes = plt.subplots(CLASSES, 5, figsize=(12, CLASSES * 2.2))
for class_idx in range(CLASSES):
    class_mask   = y_train == class_idx
    class_images = X_train[class_mask]
    indices      = np.linspace(0, len(class_images) - 1, 5, dtype=int)
    for col, img_idx in enumerate(indices):
        axes[class_idx, col].imshow(class_images[img_idx])
        axes[class_idx, col].axis('off')
    # Class name as row label on the first column only
    axes[class_idx, 0].set_title(CLASS_NAMES[class_idx], fontsize=9,
                                  loc='left', pad=4)

plt.suptitle('Synthetic X_train — 5 Samples per Class (DDIM EuroSAT)', fontsize=12)
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, 'synthetic_samples_per_class.png'),
            dpi=150, bbox_inches='tight')
plt.show()
print("Saved: synthetic_samples_per_class.png")
```

```python
# Cell 4b — Additional random sample grid for visual verification
# different seed from SEED=42 so no cell needed basically
fig, axes = plt.subplots(CLASSES, 5, figsize=(12, CLASSES * 2.2))

rng = np.random.default_rng()  # no seed — different each run
for class_idx in range(CLASSES):
    class_mask   = y_train == class_idx
    class_images = X_train[class_mask]
    indices      = rng.choice(len(class_images), size=5, replace=False)
    for col, img_idx in enumerate(indices):
        axes[class_idx, col].imshow(class_images[img_idx])
        axes[class_idx, col].axis('off')
    axes[class_idx, 0].set_title(CLASS_NAMES[class_idx], fontsize=9,
                                  loc='left', pad=4)
plt.suptitle('Synthetic X_train — Random Samples per Class (DDIM EuroSAT)', fontsize=12)
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, 'synthetic_samples_per_class_random.png'),
            dpi=150, bbox_inches='tight')
plt.show()
print("Saved: synthetic_samples_per_class_random.png")
```

```python
# Cell L1 — L1 Nearest Neighbor Distance (synthetic train vs real train)
# For each class: sample N_SAMPLES synthetic images, find closest real
# training image by L1 distance within the same class.
# Demonstrates the model is generalizing, not memorizing.
# Reference: Foster, D. (2023). Generative Deep Learning (2nd ed.). O'Reilly.

N_SAMPLES = 5

def l1_distance(img1, img2):
    return np.mean(np.abs(img1.astype("float32") - img2.astype("float32")))

real_data  = np.load(REAL_PATH)
X_real_tr  = real_data["X_train"]
y_real_tr  = real_data["y_train"]

l1_results = []
fig, axes  = plt.subplots(CLASSES * 2, N_SAMPLES,
                           figsize=(N_SAMPLES * 2.5, CLASSES * 5))

rng = np.random.default_rng(SEED)

for class_idx, class_name in enumerate(CLASS_NAMES):
    synth_mask     = y_train == class_idx
    synth_images   = X_train[synth_mask]
    sample_indices = rng.choice(len(synth_images), size=N_SAMPLES, replace=False)
    sampled_synth  = synth_images[sample_indices]

    real_mask    = y_real_tr == class_idx
    real_images  = X_real_tr[real_mask]

    class_l1_distances = []
    nearest_reals      = []

    for synth_img in sampled_synth:
        distances = [l1_distance(synth_img, real_img) for real_img in real_images]
        min_idx   = np.argmin(distances)
        min_dist  = distances[min_idx]
        nearest_reals.append((real_images[min_idx], min_dist))
        class_l1_distances.append(min_dist)

    synth_row = class_idx * 2
    real_row  = class_idx * 2 + 1

    # Top row — synthetic images
    for col in range(N_SAMPLES):
        axes[synth_row, col].imshow(sampled_synth[col])
        axes[synth_row, col].axis("off")
        if col == 0:
            axes[synth_row, col].text(-0.1, 0.5, f"{class_name}\n(Synthetic)",
                                      transform=axes[synth_row, col].transAxes,
                                      fontsize=8, va="center", ha="right",
                                      rotation=90)

    # Bottom row — nearest real neighbors
    for col, (nearest, dist) in enumerate(nearest_reals):
        axes[real_row, col].imshow(nearest)
        axes[real_row, col].axis("off")
        axes[real_row, col].set_title(f"L1={dist:.3f}", fontsize=7)
        if col == 0:
            axes[real_row, col].text(-0.1, 0.5, f"{class_name}\n(Real)",
                                      transform=axes[real_row, col].transAxes,
                                      fontsize=8, va="center", ha="right",
                                      rotation=90)

    mean_class_l1 = np.mean(class_l1_distances)
    n_synth       = int(np.sum(synth_mask))
    l1_results.append({
        "class":             class_name,
        "n_samples":         n_synth,
        "mean_l1_distance":  round(mean_class_l1, 4)
    })
    print(f"  {class_name:<25} mean L1: {mean_class_l1:.4f}  (n={n_synth})")

plt.suptitle("L1 Nearest Neighbor — Synthetic vs Real Train (DDIM EuroSAT)",
             fontsize=12)
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, "l1_nearest_neighbor.png"),
            dpi=150, bbox_inches="tight")
plt.show()
print("Saved: l1_nearest_neighbor.png")

# Weighted overall mean — weights by number of synthetic images per class
weighted_sum  = sum(r["mean_l1_distance"] * r["n_samples"] for r in l1_results)
total_samples = sum(r["n_samples"] for r in l1_results)
weighted_overall = weighted_sum / total_samples

l1_results.append({
    "class":            "Overall (weighted)",
    "n_samples":        total_samples,
    "mean_l1_distance": round(weighted_overall, 4)
})

print(f"\nOverall weighted mean L1: {weighted_overall:.4f}")

l1_df = pd.DataFrame(l1_results)
l1_df.to_csv(os.path.join(SAVE_DIR, "l1_distances.csv"), index=False)
print("Saved: l1_distances.csv")
```
