```python
# ── Cell 1: Imports & Parameters ─────────────────────────────────────────────
# Paper replication: cnn-dropout-3 from Bento (2021)
# Adadelta optimizer, 12 epochs, batch_size=128, no hyperparameter search

import os
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, callbacks
from sklearn.metrics import classification_report
import gc

# Reproducibility
SEED = 42
tf.random.set_seed(SEED)
np.random.seed(SEED)

# Data
IMAGE_SIZE = 28
CHANNELS   = 1   # grayscale
CLASSES    = 10

CLASS_NAMES = [
    'T-shirt/Top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle Boot'
]

# Training — exactly as in paper
BATCH_SIZE = 128
EPOCHS     = 12

# Paths
PROJECT_DIR  = '/path/to/thesis_project'
DATA_DIR     = PROJECT_DIR
REAL_PATH    = os.path.join(DATA_DIR, 'data', 'processed', 'seed_42','fashion_mnist.npz')
# Point to cGAN or DDIM synthetic data as needed
SYNTH_PATH   = os.path.join(DATA_DIR, 'data', 'synthetic', 'seed_42', 'ddim',
                             'fashion_mnist', 'fashion_mnist_synthetic.npz')
RESULTS_DIR  = os.path.join(PROJECT_DIR, 'results', 'cnn', 'fashion_mnist', 'paper')
os.makedirs(RESULTS_DIR, exist_ok=True)
```

```python
# ── Cell 2: Data Loading ──────────────────────────────────────────────────────

# Real data
real_data = np.load(REAL_PATH)
X_train_real = real_data['X_train']   # uint8, (48000, 28, 28, 1)
y_train_real = real_data['y_train']
X_val_real   = real_data['X_val']     # uint8, (12000, 28, 28, 1)
y_val_real   = real_data['y_val']
X_test       = real_data['X_test']    # uint8, (10000, 28, 28, 1)
y_test       = real_data['y_test']

# Synthetic data
synth_data   = np.load(SYNTH_PATH)
X_train_synth = synth_data['X_train']
y_train_synth = synth_data['y_train']
X_val_synth   = synth_data['X_val']
y_val_synth   = synth_data['y_val']

print(f'Real   — train: {X_train_real.shape}, val: {X_val_real.shape}, test: {X_test.shape}')
print(f'Synth  — train: {X_train_synth.shape}, val: {X_val_synth.shape}')

# Merge real train + val for final training (no HP search in this notebook)
X_real_merged = np.concatenate([X_train_real, X_val_real], axis=0)
y_real_merged = np.concatenate([y_train_real, y_val_real], axis=0)

X_synth_merged = np.concatenate([X_train_synth, X_val_synth], axis=0)
y_synth_merged = np.concatenate([y_train_synth, y_val_synth], axis=0)

print(f'Real merged:  {X_real_merged.shape}')
print(f'Synth merged: {X_synth_merged.shape}')
```

```python
# ── Cell 3: Preprocessing ─────────────────────────────────────────────────────
# Normalize uint8 [0,255] -> float32 [0,1] as per paper

def preprocess(X):
    return X.astype('float32') / 255.0

X_real_merged_n  = preprocess(X_real_merged)
X_synth_merged_n = preprocess(X_synth_merged)
X_test_n         = preprocess(X_test)
```

```python
# ── Cell 4: Model Architecture ────────────────────────────────────────────────
# cnn-dropout-3 from Bento (2021), Figure 4
# Two conv blocks (Conv2D -> MaxPool -> Dropout) followed by
# two dense layers with dropout, output softmax layer.
# Dropout values: 0.18 (conv blocks), 0.25 (dense layers)
# 44,426 trainable parameters

def build_model():
    model = models.Sequential([
        layers.Input(shape=(IMAGE_SIZE, IMAGE_SIZE, CHANNELS)),

        # Block 1
        layers.Conv2D(6, kernel_size=5, activation='relu'),
        layers.MaxPooling2D(pool_size=2, strides=2),
        layers.Dropout(0.18),

        # Block 2
        layers.Conv2D(16, kernel_size=5, activation='relu'),
        layers.MaxPooling2D(pool_size=2, strides=2),
        layers.Dropout(0.18),

        # Dense head
        layers.Flatten(),
        layers.Dense(120, activation='relu'),
        layers.Dropout(0.25),
        layers.Dense(84, activation='relu'),
        layers.Dropout(0.25),
        layers.Dense(CLASSES, activation='softmax'),
    ], name='cnn_dropout_3')
    return model

# Verify parameter count
test_model = build_model()
test_model.summary()
tf.keras.backend.clear_session()
del test_model
```

```python
# ── Cell 5: Training & Evaluation Function ────────────────────────────────────

def run_condition(X_train, y_train, condition_name):
    """Train on X_train/y_train, evaluate on real test set.
    No validation set used — paper trains on full available training data."""

    print(f"\n{'='*60}")
    print(f"Condition: {condition_name}")
    print(f"Train size: {X_train.shape}")
    print(f"{'='*60}")

    # Clear session — ensures no leakage from previous condition
    tf.keras.backend.clear_session()
    gc.collect()

    model = build_model()
    model.compile(
        optimizer=optimizers.Adadelta(),   # paper default: lr=1.0, rho=0.95
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    history = model.fit(
        X_train, y_train,
        batch_size=BATCH_SIZE,
        epochs=EPOCHS,
        verbose=1
    )

    # Evaluate on real test set
    test_loss, test_acc = model.evaluate(X_test_n, y_test, verbose=0)
    print(f'\nTest accuracy: {test_acc:.4f}  |  Test loss: {test_loss:.4f}')

    # Per-class metrics
    y_pred = np.argmax(model.predict(X_test_n, verbose=0), axis=1)
    report = classification_report(
        y_test, y_pred,
        target_names=CLASS_NAMES,
        digits=4
    )
    print('\nClassification Report:')
    print(report)

    # Save results
    condition_dir = os.path.join(RESULTS_DIR, condition_name)
    os.makedirs(condition_dir, exist_ok=True)

    # Save history
    pd.DataFrame(history.history).to_csv(
        os.path.join(condition_dir, 'history.csv'), index_label='epoch'
    )

    # Save classification report
    report_dict = classification_report(
        y_test, y_pred,
        target_names=CLASS_NAMES,
        digits=4,
        output_dict=True
    )
    pd.DataFrame(report_dict).transpose().to_csv(
        os.path.join(condition_dir, 'classification_report.csv')
    )

    # Plot training curves
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].plot(history.history['loss'], label='train loss')
    axes[0].set_title('Loss')
    axes[0].set_xlabel('Epoch')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)

    axes[1].plot(history.history['accuracy'], label='train accuracy')
    axes[1].set_title('Accuracy')
    axes[1].set_xlabel('Epoch')
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)

    plt.suptitle(f'Training Curves — {condition_name} (Paper Replication)', fontsize=12)
    plt.tight_layout()
    plt.savefig(os.path.join(condition_dir, 'training_curves.png'),
                dpi=150, bbox_inches='tight')
    plt.close()

    print(f'Results saved to: {condition_dir}')

    return {
        'condition':   condition_name,
        'test_acc':    round(test_acc, 4),
        'test_loss':   round(test_loss, 4),
    }
```

```python
# ── Cell 6: Run Both Conditions ───────────────────────────────────────────────
# Each condition starts from scratch with clear_session()

results = []

# Condition 1: Real Only
r = run_condition(X_real_merged_n, y_real_merged, 'real_only')
results.append(r)

# Condition 2: Synth Only
r = run_condition(X_synth_merged_n, y_synth_merged, 'synth_only')
results.append(r)

# Summary
print(f"\n{'='*60}")
print('Summary — FashionMNIST CNN Paper Replication')
print(f"{'='*60}")
summary_df = pd.DataFrame(results)
print(summary_df.to_string(index=False))
summary_df.to_csv(os.path.join(RESULTS_DIR, 'summary.csv'), index=False)
print(f'\nSummary saved to: {RESULTS_DIR}/summary.csv')
```
