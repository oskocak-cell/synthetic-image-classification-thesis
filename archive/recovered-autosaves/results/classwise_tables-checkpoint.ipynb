```python
import os
import pandas as pd
import matplotlib.pyplot as plt

BASE = os.path.join(os.path.expanduser('~'), 'Desktop', 'THESIS', 'thesis_project', 'results')

model_folders   = {'CNN': 'cnn', 'RF': 'rf', 'CNN-head': 'cnn_head', 'RF-features': 'rf_features'}
input_folders   = {'Real': 'real_only', 'DDIM': 'synth_only_ddim', 'CWGAN-GP': 'synth_only_cgan'}
dataset_folders = {'EuroSAT': 'eurosat', 'Fashion-MNIST': 'fashion_mnist', 'PathMNIST': 'path_mnist'}

MODELS       = ['CNN', 'RF', 'CNN-head', 'RF-features']
INPUTS       = ['Real', 'DDIM', 'CWGAN-GP']
SUMMARY_ROWS = ['accuracy', 'macro avg', 'weighted avg']

def load_report(model, dataset, input_type):
    path = os.path.join(
        BASE,
        model_folders[model],
        dataset_folders[dataset],
        input_folders[input_type],
        'classification_report.csv'
    )
    df = pd.read_csv(path, index_col=0)
    df = df[~df.index.isin(SUMMARY_ROWS)]
    return df

def plot_model_dataset_table(model, dataset, save_dir='.'):
    sample  = load_report(model, dataset, 'Real')
    classes = list(sample.index)

    header1 = ['Class'] + ['Real', '', '', 'DDIM', '', '', 'CWGAN-GP', '', '']
    header2 = ['']      + ['Prec.', 'Rec.', 'F1'] * 3

    cell_data = []
    for cls in classes:
        row = [cls]
        for inp in INPUTS:
            df = load_report(model, dataset, inp)
            row.append(f"{df.loc[cls, 'precision']:.4f}")
            row.append(f"{df.loc[cls, 'recall']:.4f}")
            row.append(f"{df.loc[cls, 'f1-score']:.4f}")
        cell_data.append(row)

    all_rows = [header1, header2] + cell_data
    n_cols   = len(header1)

    dark_header = '#2c3e50'
    cell_colours = []
    for i, row in enumerate(all_rows):
        if i < 2:
            cell_colours.append([dark_header] * n_cols)
            continue
        row_bg = '#f2f4f6' if i % 2 == 0 else '#ffffff'
        row_colors = [row_bg]
        for val in row[1:]:
            try:
                v = float(val)
                if v >= 0.80:
                    row_colors.append('#d4edda')
                elif v < 0.50:
                    row_colors.append('#f8d7da')
                else:
                    row_colors.append(row_bg)
            except:
                row_colors.append(row_bg)
        cell_colours.append(row_colors)

    fig_height = max(5, len(classes) * 0.45 + 2)
    fig, ax = plt.subplots(figsize=(14, fig_height))
    ax.axis('off')

    table = ax.table(
        cellText=all_rows,
        cellColours=cell_colours,
        cellLoc='center',
        loc='center'
    )
    table.auto_set_font_size(False)
    table.set_fontsize(8)
    table.scale(1, 1.6)

    for (row, col), cell in table.get_celld().items():
        cell.set_linewidth(0)
        cell.visible_edges = 'horizontal'
        if row < 2:
            cell.set_text_props(color='white', fontweight='bold', fontsize=8)
            cell.set_facecolor(dark_header)
        elif col == 0:
            cell.set_text_props(fontweight='600', fontsize=8)

    ax.set_title(
        f'{dataset} \u2014 {model}: Class-wise Precision, Recall, F1',
        fontsize=12, fontweight='600', pad=14
    )

    fname = f'classwise_{dataset.lower().replace("-", "").replace(" ", "_")}_{model.lower().replace("-", "").replace(" ", "_")}.pdf'
    save_path = os.path.join(save_dir, fname)
    plt.tight_layout()
    plt.savefig(save_path, bbox_inches='tight', dpi=150)
    plt.show()
    print(f'Saved: {save_path}')

print('Setup complete.')
```

```python
# EuroSAT — all 4 models
for model in MODELS:
    plot_model_dataset_table(model, 'EuroSAT')
```

```python
# Fashion-MNIST — all 4 models
for model in MODELS:
    plot_model_dataset_table(model, 'Fashion-MNIST')
```

```python
# PathMNIST — all 4 models
for model in MODELS:
    plot_model_dataset_table(model, 'PathMNIST')
```
