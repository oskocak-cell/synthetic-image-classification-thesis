```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
```

```python
data_dir = '/path/to/thesis_project/data/raw/EuroSAT_RGB'

ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    image_size = (64, 64),
    batch_size = None,
    shuffle = False
)
```

```python
#dataset basic info

class_names = ds.class_names
num_classes = len(class_names)

print("Number of samples:", len(ds))
print("Number of classes:", num_classes)
print("Class names:", class_names)

for image, label in ds.take(1): #inspecting data format quickly
    print("Image shape:", image.shape)
    print("Image dtype:", image.dtype)
    print("Pixel range: [{}, {}]".format(int(image.numpy().min()), int(image.numpy().max())))
```

```python
# pixel range check
pixel_min = np.inf # start with very large number
pixel_max = -np.inf # start with very small number

for image, _ in ds:
    img = image.numpy()
    pixel_min = min(pixel_min, img.min())
    pixel_max = max(pixel_max, img.max())

print("dtype:", img.dtype)
print("pixel min:", pixel_min)
print("pixel max:", pixel_max)
```

```python
#class distribution - samples per class

counts = np.zeros(num_classes, dtype=int)
for _, label in ds:
    counts[int(label)] +=1

#plot
plt.figure(figsize=(10,4))
plt.bar(class_names, counts)
plt.title("eurosat class distribution")
plt.xlabel("Class")
plt.ylabel("Number of samples")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("/path/to/thesis_project/eda/eurosat_class_distribution.png",
            dpi = 300, bbox_inches = "tight")
plt.show()

for name, count in zip(class_names, counts):
    print("{}: {}".format(name, count))
```

```python
#sample images per class

n_per_class = 5
samples = {i: [] for i in range(num_classes)}

for image, label in ds:
    l = int(label)
    if len(samples[l]) < n_per_class:
        samples[l].append(image.numpy().astype(np.uint8))
    if all(len(v) == n_per_class for v in samples.values()):
        break

#plot

fig, axes = plt.subplots(num_classes, n_per_class, figsize=(10, 20))
for i in range(num_classes):
    for j in range(n_per_class):
        axes[i, j].imshow(samples[i][j])
        axes[i, j].axis("off")
        if j == 0:
            axes[i, j].set_title(class_names[i], fontsize = 9)

plt.suptitle("eurosat sample images per class", y=1.01)
plt.tight_layout()
plt.savefig("/path/to/thesis_project/eda/eurosat_samples.png",
            dpi = 300, bbox_inches = "tight")
plt.show()
```

```python
#pixel statistics per channel

channel_sum = np.zeros(3)
channel_sq_sum = np.zeros(3)
n_pixels = 0

for image, _ in ds:
    img = image.numpy().astype(np.float32) / 255.0
    channel_sum +=img.reshape(-1, 3).sum(axis = 0)
    channel_sq_sum +=(img.reshape(-1, 3) **2).sum(axis = 0)
    n_pixels +=img.shape[0]*img.shape[1]

mean = channel_sum / n_pixels
std = np.sqrt(channel_sq_sum / n_pixels - mean**2)

for i, channel in enumerate(["R", "G", "B"]):
    print("channel {}: mean = {:.4f}, std = {:.4f}".format(channel, mean[i], std[i]))

#plot

all_pixels = {0: [], 1: [], 2: []}

for image, _ in ds:
    img = image.numpy().astype(np.float32) / 255.0
    for c in range(3):
        all_pixels[c].append(img[:, :, c].flatten())

for c in range(3):
    all_pixels[c] = np.concatenate(all_pixels[c])

colors = ["red", "green", "blue"]
channel_names = ["R", "G", "B"]

plt.figure(figsize = (10, 4))

for c in range(3):
    plt.hist(all_pixels[c], bins = 50, alpha = 0.5, color = colors[c], label = channel_names[c])

plt.title("Eurosat pixel statistics per channel")
plt.xlabel("Pixel value")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.savefig("/path/to/thesis_project/eda/eurosat_mnist_pixel_statistics.png",
            dpi = 300, bbox_inches = "tight")
plt.show()
```
