```python
# ── Cell 1: Imports & Parameters ─────────────────────────────────────────────
# Notebook 13a: CNN Head — Hyperparameter Search
# Feature extraction with frozen MobileNetV2 (ImageNet weights)
# Search on DDIM synthetic FashionMNIST train/val features (no test set)
# Hyperparameter grid same as notebook 11a

import os
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers
import itertools
import gc

SEED = 42
tf.random.set_seed(SEED)
np.random.seed(SEED)

CLASSES    = 10
TARGET_SIZE = 224  # MobileNetV2 input size

# HP grid — same as notebook 11a
PARAM_GRID = {
    'activation': ['relu', 'softmax'],
    'optimizer':  ['adadelta', 'adam', 'sgd'],
    'dropout':    [0.25, 0.5],
}

BATCH_SIZE = 128
EPOCHS     = 50

PROJECT_DIR = '/content/drive/MyDrive/thesis_project'  # adjust to your Drive path
SYNTH_PATH  = os.path.join(PROJECT_DIR, 'data', 'synthetic', 'seed_42', 'ddim',
                            'fashion_mnist', 'fashion_mnist_synthetic.npz')
RESULTS_DIR = os.path.join(PROJECT_DIR, 'results', 'cnn_head', 'hp_search')
os.makedirs(RESULTS_DIR, exist_ok=True)
```

```python
# ── Cell 2: Data Loading ──────────────────────────────────────────────────────

synth_data = np.load(SYNTH_PATH)
X_train    = synth_data['X_train']
y_train    = synth_data['y_train']
X_val      = synth_data['X_val']
y_val      = synth_data['y_val']

print(f'Synth train: {X_train.shape}, val: {X_val.shape}')
```

```python
# ── Cell 3: Preprocessing & Feature Extraction (batched) ─────────────────────
# 1. Normalize to [0, 1]
# 2. Resize to 224x224
# 3. Repeat grayscale channel 3 times → RGB
# 4. Pass through frozen MobileNetV2 → 1280-dim feature vectors

def extract_features(X, model, batch_size=64):
    features = []
    n = len(X)
    for start in range(0, n, batch_size):
        batch = X[start:start+batch_size].astype('float32') / 255.0
        batch = tf.image.resize(batch, [TARGET_SIZE, TARGET_SIZE]).numpy()
        if batch.shape[-1] == 1:
            batch = np.repeat(batch, 3, axis=-1)
        batch = tf.keras.applications.mobilenet_v2.preprocess_input(
            batch * 255.0
        )
        features.append(model.predict(batch, verbose=0))
        if (start // batch_size) % 10 == 0:
            print(f'  {start}/{n}')
    return np.concatenate(features, axis=0)

print('Loading MobileNetV2...')
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(TARGET_SIZE, TARGET_SIZE, 3),
    include_top=False,
    pooling='avg',
    weights='imagenet'
)
base_model.trainable = False

print('Extracting train features...')
F_train = extract_features(X_train, base_model)
print('Extracting val features...')
F_val   = extract_features(X_val, base_model)

print(f'Feature shape — train: {F_train.shape}, val: {F_val.shape}')
```

```python
# ── Cell 4: Hyperparameter Search ────────────────────────────────────────────

keys   = list(PARAM_GRID.keys())
combos = list(itertools.product(*[PARAM_GRID[k] for k in keys]))
print(f'Total runs: {len(combos)}\n')

records = []

for i, values in enumerate(combos):
    params   = dict(zip(keys, values))
    run_name = f"act{params['activation']}_opt{params['optimizer']}_do{params['dropout']}"

    tf.keras.backend.clear_session()
    gc.collect()

    # Dense head only — feature extraction already done
    head = models.Sequential([
        layers.Input(shape=(1280,)),
        layers.Dense(64, activation=params['activation']),
        layers.Dropout(params['dropout']),
        layers.Dense(CLASSES, activation='softmax'),
    ], name='cnn_head')

    head.compile(
        optimizer=params['optimizer'],
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    history = head.fit(
        F_train, y_train,
        validation_data=(F_val, y_val),
        batch_size=BATCH_SIZE,
        epochs=EPOCHS,
        verbose=0
    )

    best_val_acc   = max(history.history['val_accuracy'])
    final_train_acc = history.history['accuracy'][-1]
    final_val_acc   = history.history['val_accuracy'][-1]

    records.append({
        'run_name':       run_name,
        'activation':     params['activation'],
        'optimizer':      params['optimizer'],
        'dropout':        params['dropout'],
        'best_val_acc':   round(best_val_acc, 4),
        'final_train_acc': round(final_train_acc, 4),
        'final_val_acc':  round(final_val_acc, 4),
    })

    print(f'[{i+1:02d}/{len(combos)}] {run_name:40s}  '
          f'best_val={best_val_acc:.4f}  train={final_train_acc:.4f}  val={final_val_acc:.4f}')

results_df = pd.DataFrame(records)
```

```python
# ── Cell 5: Results ───────────────────────────────────────────────────────────

results_ranked = results_df.sort_values('best_val_acc', ascending=False).reset_index(drop=True)
print('Search results ranked by best_val_acc:')
print(results_ranked.to_string(index=False))

best = results_ranked.iloc[0]
print(f'\nBest run: {best["run_name"]}')
print(f'  activation: {best["activation"]}')
print(f'  optimizer:  {best["optimizer"]}')
print(f'  dropout:    {best["dropout"]}')
print(f'  best_val_acc: {best["best_val_acc"]}')

results_df.to_csv(os.path.join(RESULTS_DIR, 'search_results.csv'), index=False)
results_ranked.to_csv(os.path.join(RESULTS_DIR, 'search_results_ranked.csv'), index=False)
```

```python
# ── Cell 6: Val Accuracy Plot ─────────────────────────────────────────────────

fig, ax = plt.subplots(figsize=(14, 5))
for record in records:
    ax.plot(record['run_name'], record['best_val_acc'], 'o')

# Line plot version
fig, ax = plt.subplots(figsize=(14, 5))
run_names  = results_ranked['run_name']
val_accs   = results_ranked['best_val_acc']
bars = ax.bar(range(len(run_names)), val_accs, color='steelblue', alpha=0.8)
bars[0].set_color('darkorange')
ax.set_xticks(range(len(run_names)))
ax.set_xticklabels(run_names, rotation=45, ha='right', fontsize=8)
ax.set_ylabel('Best Val Accuracy')
ax.set_title('Val Accuracy — All 12 CNN Head Search Runs (DDIM FashionMNIST Features)')
ax.set_ylim(0, 1)
ax.grid(True, axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(RESULTS_DIR, 'val_accuracy_all_runs.png'), dpi=150, bbox_inches='tight')
plt.close()
print(f'Results saved to: {RESULTS_DIR}')
```
