```python
# ── Cell 1: Imports & Parameters ─────────────────────────────────────────────
# Architecture: 3 Conv + 2 FC layers from [CITE SECOND ARTICLE]
# 3x Conv2D(64, 2x2) + MaxPool + Dropout(0.25), Dense(64), Dense(10, softmax)
# HP search: optimizer in {Adam, SGD} x LR in {0.001, 0.01, 0.1} = 6 combinations
# Selection: best val_accuracy
# Final training: best hyperparams on merged train+val

import os
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, callbacks
from sklearn.metrics import classification_report
import gc

# Reproducibility
SEED = 42
tf.random.set_seed(SEED)
np.random.seed(SEED)

# Data
IMAGE_SIZE = 28
CHANNELS   = 1
CLASSES    = 10

CLASS_NAMES = [
    'T-shirt/Top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle Boot'
]

# Training
BATCH_SIZE    = 128
SEARCH_EPOCHS = 50
PATIENCE      = 10

# Hyperparameter search grid
# Learning rates and optimizers from Table 11 (Bento 2021)
OPTIMIZER_OPTIONS = ['adam', 'sgd']
LR_OPTIONS        = [1e-3, 1e-2, 1e-1]

# Paths
PROJECT_DIR = '/path/to/thesis_project'
DATA_DIR    = PROJECT_DIR
REAL_PATH   = os.path.join(DATA_DIR, 'fashion_mnist.npz')
SYNTH_PATH  = os.path.join(DATA_DIR, 'data', 'synthetic', 'seed_42', 'ddim',
                            'fashion_mnist', 'fashion_mnist_synthetic.npz')
RESULTS_DIR = os.path.join(PROJECT_DIR, 'results', 'cnn', 'fashion_mnist', 'search_arch2')
os.makedirs(RESULTS_DIR, exist_ok=True)
```

```python
# ── Cell 2: Data Loading ──────────────────────────────────────────────────────

real_data     = np.load(REAL_PATH)
X_train_real  = real_data['X_train']   # (48000, 28, 28, 1) uint8
y_train_real  = real_data['y_train']
X_val_real    = real_data['X_val']     # (12000, 28, 28, 1) uint8
y_val_real    = real_data['y_val']
X_test        = real_data['X_test']    # (10000, 28, 28, 1) uint8
y_test        = real_data['y_test']

synth_data    = np.load(SYNTH_PATH)
X_train_synth = synth_data['X_train']
y_train_synth = synth_data['y_train']
X_val_synth   = synth_data['X_val']
y_val_synth   = synth_data['y_val']

print(f'Real  — train: {X_train_real.shape}, val: {X_val_real.shape}, test: {X_test.shape}')
print(f'Synth — train: {X_train_synth.shape}, val: {X_val_synth.shape}')
```

```python
# ── Cell 3: Preprocessing ─────────────────────────────────────────────────────

def preprocess(X):
    return X.astype('float32') / 255.0

X_train_real_n  = preprocess(X_train_real)
X_val_real_n    = preprocess(X_val_real)
X_train_synth_n = preprocess(X_train_synth)
X_val_synth_n   = preprocess(X_val_synth)
X_test_n        = preprocess(X_test)
```

```python
# ── Cell 4: Model Architecture ────────────────────────────────────────────────
# 3 Conv blocks + 2 FC layers from [CITE SECOND ARTICLE]
# Conv blocks: Conv2D(64, 2x2, relu) -> MaxPool(2x2) -> Dropout(0.25)
# Third conv block has no pooling layer as per article description
# FC head: Dense(64, relu) -> Dropout(0.25) -> Dense(10, softmax)
# ReLU used for all hidden layers — standard activation for CNNs
# padding='same' preserves spatial dimensions through conv layers

def build_model():
    model = models.Sequential([
        layers.Input(shape=(IMAGE_SIZE, IMAGE_SIZE, CHANNELS)),

        # Block 1
        layers.Conv2D(64, kernel_size=2, padding='same', activation='relu'),
        layers.MaxPooling2D(pool_size=2),
        layers.Dropout(0.25),

        # Block 2
        layers.Conv2D(64, kernel_size=2, padding='same', activation='relu'),
        layers.MaxPooling2D(pool_size=2),
        layers.Dropout(0.25),

        # Block 3 — no pooling as per article
        layers.Conv2D(64, kernel_size=2, padding='same', activation='relu'),
        layers.Dropout(0.25),

        # FC head
        layers.Flatten(),
        layers.Dense(64, activation='relu'),
        layers.Dropout(0.25),
        layers.Dense(CLASSES, activation='softmax'),
    ], name='cnn_3conv_2fc')
    return model

def get_optimizer(name, lr):
    if name == 'adam':
        return optimizers.Adam(learning_rate=lr)
    elif name == 'sgd':
        # Default momentum=0.0 as per Keras default
        return optimizers.SGD(learning_rate=lr)

# Verify architecture
test_model = build_model()
test_model.summary()
tf.keras.backend.clear_session()
del test_model
```

```python
# ── Cell 5: Training & Evaluation Functions ───────────────────────────────────

def run_search(X_train, y_train, X_val, y_val, condition_name):
    """Hyperparameter search for one condition.
    Returns best optimizer name, best LR, and full search results."""

    print(f"\n{'='*60}")
    print(f"Hyperparameter Search — {condition_name}")
    print(f"{'='*60}")

    search_dir = os.path.join(RESULTS_DIR, condition_name, 'search')
    os.makedirs(search_dir, exist_ok=True)
    search_results = []

    for opt_name in OPTIMIZER_OPTIONS:
        for lr in LR_OPTIONS:
            run_name = f'{opt_name}_lr{lr}'
            print(f'\n── Run: {run_name}')

            # Clear session — no leakage across runs
            tf.keras.backend.clear_session()
            gc.collect()

            model = build_model()
            model.compile(
                optimizer=get_optimizer(opt_name, lr),
                loss='sparse_categorical_crossentropy',
                metrics=['accuracy']
            )

            early_stop = callbacks.EarlyStopping(
                monitor='val_accuracy',
                patience=PATIENCE,
                restore_best_weights=True,
                mode='max'
            )

            history = model.fit(
                X_train, y_train,
                batch_size=BATCH_SIZE,
                epochs=SEARCH_EPOCHS,
                validation_data=(X_val, y_val),
                callbacks=[early_stop],
                verbose=1
            )

            best_val_acc = max(history.history['val_accuracy'])
            best_epoch   = history.history['val_accuracy'].index(best_val_acc) + 1
            print(f'   Best val_accuracy: {best_val_acc:.4f} at epoch {best_epoch}')

            # Save run history
            run_dir = os.path.join(search_dir, run_name)
            os.makedirs(run_dir, exist_ok=True)
            pd.DataFrame(history.history).to_csv(
                os.path.join(run_dir, 'history.csv'), index_label='epoch'
            )

            search_results.append({
                'run_name':     run_name,
                'optimizer':    opt_name,
                'lr':           lr,
                'best_val_acc': round(best_val_acc, 4),
                'best_epoch':   best_epoch,
            })

            pd.DataFrame(search_results).to_csv(
                os.path.join(search_dir, 'search_results.csv'), index=False
            )

    # Summary
    print(f"\n── Search Results — {condition_name}")
    for r in sorted(search_results, key=lambda x: x['best_val_acc'], reverse=True):
        print(f"   {r['run_name']:<25}  val_acc: {r['best_val_acc']:.4f}  epoch: {r['best_epoch']}")

    best_run = max(search_results, key=lambda x: x['best_val_acc'])
    print(f"\n   Best: {best_run['run_name']}  (val_acc={best_run['best_val_acc']:.4f})")

    return best_run['optimizer'], best_run['lr'], search_results


def run_final_training(X_train, y_train, opt_name, lr, condition_name, search_results):
    """Final training from scratch on merged train+val with best hyperparams."""

    print(f"\n── Final Training — {condition_name}")
    print(f"   Optimizer: {opt_name}  LR: {lr}  Train size: {X_train.shape}")

    tf.keras.backend.clear_session()
    gc.collect()

    best_run     = max(search_results, key=lambda x: x['best_val_acc'])
    final_epochs = best_run['best_epoch']
    print(f"   Training for {final_epochs} epochs (from best search run)")

    model = build_model()
    model.compile(
        optimizer=get_optimizer(opt_name, lr),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    history = model.fit(
        X_train, y_train,
        batch_size=BATCH_SIZE,
        epochs=final_epochs,
        verbose=1
    )

    # Evaluate on real test set
    test_loss, test_acc = model.evaluate(X_test_n, y_test, verbose=0)
    print(f'\n   Test accuracy: {test_acc:.4f}  |  Test loss: {test_loss:.4f}')

    # Per-class metrics
    y_pred  = np.argmax(model.predict(X_test_n, verbose=0), axis=1)
    report  = classification_report(
        y_test, y_pred, target_names=CLASS_NAMES, digits=4
    )
    print('\n   Classification Report:')
    print(report)

    # Save
    final_dir = os.path.join(RESULTS_DIR, condition_name, 'final')
    os.makedirs(final_dir, exist_ok=True)

    pd.DataFrame(history.history).to_csv(
        os.path.join(final_dir, 'history.csv'), index_label='epoch'
    )

    report_dict = classification_report(
        y_test, y_pred, target_names=CLASS_NAMES, digits=4, output_dict=True
    )
    pd.DataFrame(report_dict).transpose().to_csv(
        os.path.join(final_dir, 'classification_report.csv')
    )

    # Training curves
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].plot(history.history['loss'], label='train loss')
    axes[0].set_title('Loss')
    axes[0].set_xlabel('Epoch')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)

    axes[1].plot(history.history['accuracy'], label='train accuracy')
    axes[1].set_title('Accuracy')
    axes[1].set_xlabel('Epoch')
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)

    plt.suptitle(f'Training Curves — {condition_name} (Final)', fontsize=12)
    plt.tight_layout()
    plt.savefig(
        os.path.join(final_dir, 'training_curves.png'),
        dpi=150, bbox_inches='tight'
    )
    plt.close()

    print(f'   Results saved to: {final_dir}')

    return {
        'condition': condition_name,
        'optimizer': opt_name,
        'lr':        lr,
        'test_acc':  round(test_acc, 4),
        'test_loss': round(test_loss, 4),
    }
```

```python
# ── Cell 6: Run Real Only Condition ───────────────────────────────────────────

best_opt_real, best_lr_real, search_results_real = run_search(
    X_train_real_n, y_train_real,
    X_val_real_n,   y_val_real,
    condition_name='real_only'
)

X_real_merged_n = np.concatenate([X_train_real_n, X_val_real_n], axis=0)
y_real_merged   = np.concatenate([y_train_real,   y_val_real],   axis=0)

result_real = run_final_training(
    X_real_merged_n, y_real_merged,
    best_opt_real, best_lr_real,
    condition_name='real_only',
    search_results=search_results_real
)
```

```python
# ── Cell 7: Run Synth Only Condition ──────────────────────────────────────────

best_opt_synth, best_lr_synth, search_results_synth = run_search(
    X_train_synth_n, y_train_synth,
    X_val_synth_n,   y_val_synth,
    condition_name='synth_only'
)

X_synth_merged_n = np.concatenate([X_train_synth_n, X_val_synth_n], axis=0)
y_synth_merged   = np.concatenate([y_train_synth,   y_val_synth],   axis=0)

result_synth = run_final_training(
    X_synth_merged_n, y_synth_merged,
    best_opt_synth, best_lr_synth,
    condition_name='synth_only',
    search_results=search_results_synth
)
```

```python
# ── Cell 8: Summary ───────────────────────────────────────────────────────────

print(f"\n{'='*60}")
print('Final Summary — FashionMNIST CNN 3Conv2FC with HP Search')
print(f"{'='*60}")

summary_df = pd.DataFrame([result_real, result_synth])
print(summary_df.to_string(index=False))
summary_df.to_csv(os.path.join(RESULTS_DIR, 'summary.csv'), index=False)
print(f'\nSummary saved to: {RESULTS_DIR}/summary.csv')
```
